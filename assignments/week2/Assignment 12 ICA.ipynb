```python
import pandas as pd import numpy as np # Step 1: Load raw messy dataset df\_raw = pd.read\_csv('teaching\_prices\_messy.csv') print("=== BEFORE CLEANING AUDIT ===") print("Total Rows:", len(df\_raw)) print("\\nMissing Values per Column:") print(df\_raw.isnull().sum()) print("\\nData Types:") print(df\_raw.dtypes) # Step 2: Clean on an explicit copy df\_clean = df\_raw.copy() # A. Remove exact duplicate rows df\_clean = df\_clean.drop\_duplicates() # B. Clean and standardize categorical string columns if 'item' in df\_clean.columns: df\_clean['item'] = df\_clean['item'].astype(str).str.strip().str.title() if 'city' in df\_clean.columns: df\_clean['city'] = df\_clean['city'].astype(str).str.strip().str.title() if 'unit' in df\_clean.columns: df\_clean['unit'] = df\_clean['unit'].astype(str).str.strip().str.lower() # C. Clean 'price' column (remove symbols, cast to numeric float) if 'price' in df\_clean.columns: if df\_clean['price'].dtype == 'object': df\_clean['price'] = ( df\_clean['price'] .astype(str) .str.replace('Rs.', '', regex=False) .str.replace('Rs', '', regex=False) .str.replace(',', '', regex=False) .str.strip() ) df\_clean['price'] = pd.to\_numeric(df\_clean['price'], errors='coerce') # D. Filter out impossible prices (negative or zero) df\_clean = df\_clean[df\_clean['price'] &gt; 0] # E. Parse date column to datetime if 'date' in df\_clean.columns: df\_clean['date'] = pd.to\_datetime(df\_clean['date'], errors='coerce') # F. Drop rows with null values in core fields df\_clean = df\_clean.dropna(subset=['item', 'price', 'date']) # Step 3: Before vs After comparison audit table print("\\n=== DATA AUDIT COMPARISON (BEFORE vs AFTER) ===") comparison\_metrics = { "Metric": [ "Total Rows", "Total Columns", "Missing Price Values", "Duplicate Rows", "Distinct Items", "Distinct Cities" ], "Before Cleaning": [ len(df\_raw), df\_raw.shape[1], df\_raw['price'].isnull().sum() if 'price' in df\_raw.columns else "N/A", df\_raw.duplicated().sum(), df\_raw['item'].nunique() if 'item' in df\_raw.columns else "N/A", df\_raw['city'].nunique() if 'city' in df\_raw.columns else "N/A" ], "After Cleaning": [ len(df\_clean), df\_clean.shape[1], df\_clean['price'].isnull().sum(), df\_clean.duplicated().sum(), df\_clean['item'].nunique(), df\_clean['city'].nunique() ] } comparison\_df = pd.DataFrame(comparison\_metrics) print(comparison\_df.to\_string(index=False)) print("\\n=== AFTER CLEANING SUMMARY STATISTICS ===") print(df\_clean.describe())
```
